```python
from collections.abc import Sequence
from random import sample
from itertools import permutations

TOTAL_ITEMS = 15
TOTAL_ATTRIBUTES = 20

class Item:
    _title: str
    _attributes: tuple[bool, ...]
    def __init__(self, title: str, attributes: Sequence[bool]):
        self._title = title
        self._attributes = tuple(attributes)

    def distance(self, other: "Item") -> int:
        return sum(a != b for a, b in zip(self._attributes, other._attributes))

elements = []

for idx in range(TOTAL_ITEMS):
    title = f"Item {idx}"
    elements.append(Item(title, sample([True, False], k=TOTAL_ATTRIBUTES, counts=[TOTAL_ATTRIBUTES, TOTAL_ATTRIBUTES])))

for elem in elements:
    print(elem)

def evaluate_score(arrangement: Sequence[Item]):
    size = len(arrangement)
    return sum(arrangement[i].distance(arrangement[(i + 1) % size]) for i in range(size))

def mutate(arrangement: Sequence[Item]) -> list[Item]:
    idx1, idx2 = sample(range(len(arrangement)), k=2)
    modified = list(arrangement)
    modified[idx1] = arrangement[idx2]
    modified[idx2] = arrangement[idx1]
    return modified

current_state = elements[:]
optimal_score = evaluate_score(current_state)
print("Initial fitness", optimal_score)

for _ in range(10000):
    candidate_state = mutate(current_state)
    candidate_score = evaluate_score(candidate_state)

    if candidate_score > optimal_score:
        optimal_score = candidate_score
        current_state = candidate_state

print("Best solution:", [e._title for e in current_state])
print("Best fitness:", optimal_score)
```
